```python

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Read external dataset
data = pd.read_csv("products.csv")

# Set theme
sns.set_theme(style="whitegrid")

# Create subplots
fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(12, 5)
)

# Box plot
sns.boxplot(
    data=data,
    x="Product Category",
    y="price",
    hue="Product Category",
    palette=["red", "blue", "green", "orange"],
    legend=False,
    ax=axes[0]
)

axes[0].set_title("Product Price Distribution (Box Plot)")
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Price")

# Violin plot
sns.violinplot(
    data=data,
    x="Product Category",
    y="price",
    hue="Product Category",
    palette=["red", "blue", "green", "orange"],
    legend=False,
    ax=axes[1]
)

axes[1].set_title("Product Price Distribution (Violin Plot)")
axes[1].set_xlabel("Product Category")
axes[1].set_ylabel("Price")

# Main title
fig.suptitle("Product Price Analysis")

plt.tight_layout()

# Save output
plt.savefig("product_analysis.png", dpi=300)

# Display output
plt.show()
```
